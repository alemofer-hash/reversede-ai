# Reversede quickstart

This notebook uses only synthetic data. Install the checkout with `python -m pip install -e ".[sklearn]"` before running it.

```python
import pandas as pd
import reversede
from reversede import Reversede, ReversedeConfig

reversede.__version__
```

```python
frame = pd.DataFrame({
    "temperature": [12.1, 18.4, 22.0, 28.7, 31.2, 37.6],
    "pressure": [1.0, 1.2, 1.1, 1.5, 1.7, 1.9],
    "family": ["a", "a", "b", "b", "c", "c"],
    "critical_temp": [4.0, 6.1, 7.8, 10.2, 12.0, 14.5],
})
frame
```

```python
analysis = Reversede(
    frame,
    target="critical_temp",
    task="regression",
    config=ReversedeConfig(max_features=3),
)
analysis.candidates_[["column", "selected", "ranking_score"]]
```

```python
blueprint = analysis.generate_blueprint()
compile(blueprint, "reversede_blueprint.py", "exec")
{
    "selected_features": analysis.selected_features_,
    "alerts": analysis.alerts_,
    "blueprint_lines": len(blueprint.splitlines()),
}
```
