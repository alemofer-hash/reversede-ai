# Install Reversede from the private repository

Grant your GitHub account access first. In Colab Secrets, add a fine-grained, read-only token named `REVERSEDE_GITHUB_TOKEN`. The token is never printed or stored in notebook output.

```python
import base64
import os
import subprocess
import sys

from google.colab import userdata

token = userdata.get("REVERSEDE_GITHUB_TOKEN")
if not token:
    raise RuntimeError("Colab secret REVERSEDE_GITHUB_TOKEN is not configured")

basic = base64.b64encode(f"x-access-token:{token}".encode()).decode()
git_env = os.environ.copy()
git_env.update({
    "GIT_CONFIG_COUNT": "1",
    "GIT_CONFIG_KEY_0": "http.https://github.com/.extraheader",
    "GIT_CONFIG_VALUE_0": f"AUTHORIZATION: basic {basic}",
})
requirement = (
    "reversede-ai[sklearn] @ "
    "git+https://github.com/alemofer-hash/reversede-ai.git@v0.3.0-dev0"
)
subprocess.run([sys.executable, "-m", "pip", "install", requirement], env=git_env, check=True)
del token, basic, git_env
print("Reversede installed from the authorized private repository.")
```
